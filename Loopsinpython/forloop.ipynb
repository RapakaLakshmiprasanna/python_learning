for loop in python will contains 'range' which will help for the iteration process,their is an object is created that will help to iterate the loop

-> based on the situations in python , loops will be used 
like : if the end of the iteration is well known then we can use for() loop 
if not then we need to use while loop()

```python
#range 
list(range(10))
```

```python
print(list(range(1,10)))
```

```python
print(list(range(0,10,2))) # we can get even numbers from 0 to 10
print(list(range(1,10,11))) # we can get 1 number from 1 to 10 becoz gap = 11 so 1+11=12 but end of range is before 10 so we get only 1 number
print(list(range(1,11,2))) # we can get odd numbers from 1 to 10
```

```python
list(range(10,0,-1))
```

```python
for i in range(10):
    print(i, end="*")
print()
```

Example: print numbers from 0 to 9 in one line separated by `*`.

```python
print(*range(10), sep="*")
```

```python
for i in "kolkata":
    print(i)
```

```python
for i in [1,2,3]:
    print(i)
    for j in [4,5,6]:
        print(j)
```

```python
# *
# **
# ***
# ****
# ***** printing pattern using for loop
rows =int(input("Enter number of rows: "))
for i in range(rows):
    for j in range(i+1):
        print("*", end="")
    print()
```
